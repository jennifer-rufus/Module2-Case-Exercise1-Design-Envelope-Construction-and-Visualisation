FMECA stands for Failure Modes, Effects and Criticality Analysis. It is a systematic engineering technique used to identify potential equipment failures, determine their causes and consequences, evaluate their risks, and recommend preventive actions. A subsea Christmas tree is an assembly of valves, actuators, seals, connectors and control interfaces that controls oil and gas flow from a subsea well.

FMECA rating system: Severity (S), Occurrence (O) and Detection (D) are rated from 1 to 10. A higher Detection score means the failure is harder to detect.

Q3. Implement at least two of the physics models in Python and validate against hand calculations.



Model 1: Barlow's pressure model
Barlow's equation estimates the pressure capacity of a thin-walled cylindrical pipe. Barlow’s formula is used to estimate the internal pressure capacity of a pipe based on its outside diameter, wall thickness, and material yield strength.

            P = 2st / D

Where S is yield strebght, t is wall thickness, and D is outside diameter

```python
def barlow_pressure(S, t, D):
    if S <= 0 or t <= 0 or D <= 0:
        raise ValueError("All inputs must be positive.")

    if 2 * t >= D:
        raise ValueError("Thin-wall assumption is not satisfied.")

    return (2 * S * t) / D


S = 65000       # psi
t = 0.50        # inches
D = 10.75       # inches

python_pressure = barlow_pressure(S, t, D)

#Hand calculation
hand_pressure = (2 * 65000 * 0.50) / 10.75

print("Python result:", round(python_pressure, 2), "psi")
print("Hand calculation:", round(hand_pressure, 2), "psi")

assert abs(python_pressure - hand_pressure) < 0.01

print("Barlow model validated successfully.")
```

Model 2: Corrosion rate and remaining useful life.
The corrosion rate is estimated from the change in wall thickness over time. the objective of this model estimates the corrosion rate of a subsea pipe and its remaining useful life using measured wall thickness at two different times.

            CR = t0 - t1 / T

The remaining useful life under a constant corrosion-rate assumption is:

            RUL = t1 - tmin / CR

```python
def calculate_corrosion_rate(initial, current, years):
    if years <= 0:
        raise ValueError("Years must be positive.")

    if initial <= current:
        raise ValueError("Initial thickness must exceed current thickness.")

    return (initial - current) / years


def calculate_rul(current, minimum, rate):
    if rate < 0:
        raise ValueError("Corrosion rate cannot be negative.")

    if current <= minimum:
        return 0.0

    if rate == 0:
        return float("inf")

    return (current - minimum) / rate


initial = 12.0
current = 10.8
years = 2.0
minimum = 8.0

python_rate = calculate_corrosion_rate(
    initial, current, years
)

python_rul = calculate_rul(
    current, minimum, python_rate
)

# Hand calculations
hand_rate = (12.0 - 10.8) / 2.0
hand_rul = (10.8 - 8.0) / 0.60

print("Python corrosion rate:", round(python_rate, 2), "mm/year")
print("Hand corrosion rate:", round(hand_rate, 2), "mm/year")

print("Python RUL:", round(python_rul, 2), "years")
print("Hand RUL:", round(hand_rul, 2), "years")

assert abs(python_rate - hand_rate) < 1e-9
assert abs(python_rul - hand_rul) < 1e-9

print("Both physics models validated successfully.")
```
